# Quickstart voice session
Complete [setup](../../instructions/00_setup.md) first. This is a **live Azure** exercise, not a mock. It uses the microphone on the kernel machine and may incur charges. Use a headset and synthetic speech only. Audio stays in memory; clear outputs before sharing.

Run each cell in order. When recording starts, say: **Explain how a rainbow forms in two short sentences.**

```python
from pathlib import Path
import sys
from dotenv import load_dotenv
from IPython.display import Audio, display

root = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / 'src/shared/realtime.py').is_file())
sys.path.insert(0, str(root / 'src'))
from shared.realtime import Settings, connect, record, exchange, wav_bytes
load_dotenv(root / '.env', override=False)
settings = Settings.from_env()
_ = settings.url
print('Local configuration validated; no service connection attempted yet.')
```

```python
async with connect(settings, 'Explain science clearly in at most two short sentences.') as session:
    print('Live session configured. Speak now: recording for 5 seconds.')
    microphone_audio = await record(5)
    response_audio = await exchange(session, microphone_audio)
    events = list(session.events)
display(Audio(wav_bytes(response_audio)))
events
```
