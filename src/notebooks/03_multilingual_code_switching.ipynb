# Multilingual conversations and code-switching
Complete [setup](../../instructions/00_setup.md) and quickstart first. This is a **live Azure** microphone exercise. It compares English, Spanish, and a mixed-language utterance in separate fresh sessions, without claiming guaranteed language or pronunciation quality. Each run may incur charges.

Run the final cell three times, changing `case_index` to 0, 1, then 2. Read the displayed prompt aloud during the five-second recording. Listen and record the language actually used. Do not confuse language steering with a pronunciation-customization feature.

```python
from pathlib import Path
import json
import sys
from dotenv import load_dotenv
from IPython.display import Audio, display

root = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / 'src/shared/realtime.py').is_file())
sys.path.insert(0, str(root / 'src'))
from shared.realtime import Settings, connect, record, exchange, wav_bytes
load_dotenv(root / '.env', override=False)
settings = Settings.from_env()
_ = settings.url
cases = json.loads((root / 'data/prompts/multilingual.json').read_text())
print('Synthetic prompts loaded locally. No live result yet.')
```

```python
case_index = 0
case = cases[case_index]
print('Read aloud when recording starts:', case['utterance'])
async with connect(settings, 'Explain science briefly. Follow the language requested by the speaker, including language changes.') as session:
    print('Live session configured. Speak now: recording for 5 seconds.')
    microphone_audio = await record(5)
    response_audio = await exchange(session, microphone_audio)
    events = list(session.events)
display(Audio(wav_bytes(response_audio)))
print('Requested behavior:', case['expected_behavior'])
events
```
